```python
import os
import numpy as np
import pandas as pd

red_df = pd.read_csv("https://raw.githubusercontent.com/ymudgal24/ymudgal24.github.io/main/Red.csv")
rose_df = pd.read_csv("https://raw.githubusercontent.com/ymudgal24/ymudgal24.github.io/main/Rose.csv")
white_df = pd.read_csv("https://raw.githubusercontent.com/ymudgal24/ymudgal24.github.io/main/White.csv")
sparkling_df = pd.read_csv("https://raw.githubusercontent.com/ymudgal24/ymudgal24.github.io/main/Sparkling.csv")
```

```python
# prompt: create bucket of rating on 0.5 scale and give me number of wines falling into category

import pandas as pd
import numpy as np
# Assuming 'wine_df' is your DataFrame as defined in the previous code.

# Create rating buckets
bins = np.arange(0, 5.5, 0.5)  # Create bins from 0 to 5 with 0.5 intervals
labels = [f'{i}-{i+0.5}' for i in bins[:-1]] #Create labels for each bin

filtered_wine_df['rating_bucket'] = pd.cut(filtered_wine_df['Rating'], bins=bins, labels=labels, right=False)

# Count wines in each bucket
rating_counts = filtered_wine_df.groupby('rating_bucket')['Rating'].count()

rating_counts
```

```python
# prompt: Histogram → Compare price distribution for high vs. low-rated wines

import matplotlib.pyplot as plt
# Define high and low rating thresholds
high_rating_threshold = 4.5
low_rating_threshold = 3.0

# Filter wines based on rating thresholds
high_rated_wines = filtered_wine_df[filtered_wine_df['Rating'] >= high_rating_threshold]
low_rated_wines = filtered_wine_df[filtered_wine_df['Rating'] <= low_rating_threshold]


# Create histograms for high and low-rated wines
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(high_rated_wines['Price'], bins=20, color='skyblue', edgecolor='black')
plt.title('Price Distribution of High-Rated Wines')
plt.xlabel('Price')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(low_rated_wines['Price'], bins=20, color='salmon', edgecolor='black')
plt.title('Price Distribution of Low-Rated Wines')
plt.xlabel('Price')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Define rating thresholds
high_rating_threshold = 4.5
low_rating_threshold = 3.0

# Filter wines into three categories
high_rated_wines = filtered_wine_df[filtered_wine_df['Rating'] >= high_rating_threshold]
medium_rated_wines = filtered_wine_df[(filtered_wine_df['Rating'] > low_rating_threshold) &
                                      (filtered_wine_df['Rating'] < high_rating_threshold)]
low_rated_wines = filtered_wine_df[filtered_wine_df['Rating'] <= low_rating_threshold]

# Scatter plot for high, medium, and low-rated wines
plt.figure(figsize=(10, 6))

plt.scatter(high_rated_wines['Rating'], high_rated_wines['Price'],
            color='blue', alpha=0.6, label='High-Rated Wines (>= 4.5)')
plt.scatter(medium_rated_wines['Rating'], medium_rated_wines['Price'],
            color='green', alpha=0.6, label='Medium-Rated Wines (3.0 - 4.5)')
plt.scatter(low_rated_wines['Rating'], low_rated_wines['Price'],
            color='red', alpha=0.6, label='Low-Rated Wines (<= 3.0)')

# Labels and title
plt.xlabel('Rating')
plt.ylabel('Price')
plt.title('Price vs. Rating for High, Medium, and Low-Rated Wines')
plt.legend()
plt.grid(True)

plt.show()
```

```python
# prompt: use the two graphs above to show two world maps of wines by country showing prices and ratings in colour and remove any prices above 3000. reset the scale for prices to the mean values by country

import plotly.express as px

# Calculate the mean price for each country
mean_prices_by_country = filtered_wine_df.groupby('Country')['Price'].mean().reset_index()

# Create the first world map for prices
fig_price = px.choropleth(mean_prices_by_country, locations='Country', locationmode='country names',
                           color='Price', hover_name='Country',
                           title='Average Wine Price by Country (<= $3000)',
                           color_continuous_scale=px.colors.sequential.Plasma)
fig_price.show()

# Calculate the mean rating for each country
mean_ratings_by_country = filtered_wine_df.groupby('Country')['Rating'].mean().reset_index()

# Create the second world map for ratings
fig_rating = px.choropleth(mean_ratings_by_country, locations='Country', locationmode='country names',
                            color='Rating', hover_name='Country',
                            title='Average Wine Rating by Country',
                            color_continuous_scale=px.colors.sequential.Viridis)
fig_rating.show()
```

```python
# prompt: create a new dataframe where you remove all columns with year listed as N.V. and price is less than 3000 and visualise a correlation matrix between rating, number of ratings, price, and year. also explain the relationship between price and year. list the average price for year

import pandas as pd
import matplotlib.pyplot as plt
# Remove rows where 'Year' is 'N.V.' and 'Price' is less than 3000
# filtered_wine_df = wine_df[(wine_df['Year'] != 'N.V.') | (wine_df['Price'] >= 3000)]

# Select relevant columns for correlation and handle missing values
columns_for_correlation = ['Rating', 'NumberOfRatings', 'Price', 'Year']
correlation_df = filtered_wine_df[columns_for_correlation].dropna()

# Convert 'Year' to numeric, coercing errors to NaN
correlation_df['Year'] = pd.to_numeric(correlation_df['Year'], errors='coerce')
correlation_df = correlation_df.dropna(subset=['Year'])

# Calculate the correlation matrix
correlation_matrix = correlation_df.corr()

# Visualize the correlation matrix
plt.figure(figsize=(10, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Wine Features')
plt.show()

```

```python
# prompt: create a histogram to show the number of wines and reviews by year, use dataset filtered_wine_df

import pandas as pd
import matplotlib.pyplot as plt

# Assuming 'filtered_wine_df' is already defined and loaded

# Convert 'Year' column to numeric, handling errors
filtered_wine_df['Year'] = pd.to_numeric(filtered_wine_df['Year'], errors='coerce')

# Drop rows with NaN values in the 'Year' column
filtered_wine_df = filtered_wine_df.dropna(subset=['Year'])


# Create the histogram
plt.figure(figsize=(12, 6))
plt.hist(filtered_wine_df['Year'], bins=20)  # Adjust the number of bins as needed
plt.xlabel('Year')
plt.ylabel('Number of Wines')
plt.title('Distribution of Wines by Year')
plt.show()

# Create a histogram for the number of reviews per year
plt.figure(figsize=(12,6))
plt.hist(filtered_wine_df['Year'], weights = filtered_wine_df['NumberOfRatings'], bins = 20)
plt.xlabel('Year')
plt.ylabel('Number of Reviews')
plt.title('Number of Reviews by Year')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Group data by year and calculate the average rating and price
yearly_data = filtered_wine_df.groupby('Year')[['Rating', 'Price']].mean().reset_index()

# Convert Year to numpy array explicitly
years = np.array(yearly_data['Year'])

# Create the first plot for average rating
plt.figure(figsize=(10, 6))
plt.plot(years, yearly_data['Rating'].values, marker='o')
plt.title('Average Wine Rating per Year')
plt.xlabel('Year')
plt.ylabel('Average Rating')
plt.grid(True)
plt.show()

# Create the second plot for average price
plt.figure(figsize=(10, 6))
plt.plot(years, yearly_data['Price'].values, marker='o', color='orange')
plt.title('Average Wine Price per Year')
plt.xlabel('Year')
plt.ylabel('Average Price')
plt.grid(True)
plt.show()
```

```python
# prompt: For each country in the red wine dataset, create a histogram of Price, a histogram of Rating, and a scatterplot of Price vs. Rating

import pandas as pd
import matplotlib.pyplot as plt

# Assuming red_df is already defined and loaded with data
# If not, load the data first:
# red_df = pd.read_csv("Red.csv")

for country in red_df['Country'].unique():
    country_df = red_df[red_df['Country'] == country]

    # Price Histogram
    plt.figure(figsize=(8, 6))
    plt.hist(country_df['Price'], bins=10)
    plt.xlabel('Price')
    plt.ylabel('Frequency')
    plt.title(f'Price Histogram for {country} Red Wine')
    plt.show()

    # Rating Histogram
    plt.figure(figsize=(8, 6))
    plt.hist(country_df['Rating'], bins=10)
    plt.xlabel('Rating')
    plt.ylabel('Frequency')
    plt.title(f'Rating Histogram for {country} Red Wine')
    plt.show()

    # Scatterplot of Price vs. Rating
    plt.figure(figsize=(8, 6))
    plt.scatter(country_df['Price'], country_df['Rating'])
    plt.xlabel('Price')
    plt.ylabel('Rating')
    plt.title(f'Price vs. Rating Scatterplot for {country} Red Wine')
    plt.show()
```
